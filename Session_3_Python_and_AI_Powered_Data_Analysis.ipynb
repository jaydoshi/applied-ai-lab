# Part 1: Demo

### Step 1: Load the data

> Import pandas and plotly express. Load the customer file and show the first five rows.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("Session 3 - Customer Data.csv")

df.head()
```

```python
print("Rows and columns:", df.shape)
```

### Step 2: Which channel brings the most valuable customers?

> Show the number of customers and average 12-month revenue by channel, sorted, as a bar chart.

```python
by_channel = df.groupby("channel")["revenue_12m"].agg(["count", "mean"]).sort_values("mean")
by_channel
```

Visulalize the channels revenue:

```python
# Visualize the average 12-month revenue by channel
fig = px.bar(
    by_channel.reset_index(),
    x="channel",
    y="mean",
    title="Average 12-Month Revenue by Channel",
    labels={"mean": "Average Revenue ($)", "channel": "Acquisition Channel"}
)
fig.show()
```

### Step 3: What does each customer cost?

> For each channel, compute total revenue divided by total acquisition cost.

```python
totals = df.groupby("channel")[["revenue_12m", "acquisition_cost"]].sum()
totals["return_on_spend"] = totals["revenue_12m"] / totals["acquisition_cost"]
totals.sort_values("return_on_spend").round(1)
```

### Step 5: Linear regression

A **linear regression** models the **expected value** of an outcome as a linear function of **predictors**. Each **coefficient** is the expected change in the outcome for a one-unit increase in that predictor, holding the other predictors constant. Categorical predictors are encoded automatically, and each category is compared with a **reference category**. **R-squared** is the share of the outcome's variance the model explains.

**Model 1: who the customer is and how we acquired them**

> Using statsmodels formulas, fit a regression of revenue_12m on customer_type, channel, and discount_pct. Show the coefficient table and adjusted R-squared.

```python
import statsmodels.formula.api as smf

model_1 = smf.ols("revenue_12m ~ customer_type + channel + discount_pct", data=df).fit()

print("R-squared:", round(model_1.rsquared, 3), " Adjusted:", round(model_1.rsquared_adj, 3))
model_1.summary().tables[1]
```
